## 3. Cálculo Diferencial: Derivadas e Gradiente

**Exercício 6:** Defina uma função quadrática $f(x) = x^2 + 5x + 10$ e sua derivada analítica $f'(x)$. Realize 10 passos de uma otimização simples para encontrar um valor menor de $x$ partindo de $x=10$.

```python
# Ex. 6
def f(x):
  return x**2 + 5*x + 10

def df(x):
  return 2*x + 5

x = 10.0
learn_rate = 0.1

for i in range(10):
  gradiente = df(x)
  passo = learn_rate * gradiente
  x = x - passo
  print(f'Passo {i+1}: X = {x:.2f} | Gradiente = {gradiente:.2f} | Passo = {passo:.2f}')
```

```python
# Ex. 6 modificado
def f(x):
  return x**2 + 5*x + 10

def df(x):
  return 2*x + 5

x = 10.0
learn_rate = 0.1
tolerancia = 0.005
i = 0

while abs(df(x)>=tolerancia):
  i += 1
  gradiente = df(x)
  passo = learn_rate * gradiente
  x = x - passo
  print(f'Passo {i}: X = {x:.6f} | Gradiente = {gradiente:.2f} | Passo = {passo:.2f}')
  if (i>100):
    print('Limite de iterações atingido')
    break
```

## 4. Estatística e Probabilidade

**Exercício 7:** Implemente a função Softmax. Ela é fundamental para converter scores (logits) em uma distribuição de probabilidade onde a soma é 1. Use o vetor `[2.0, 1.0, 0.1]` como teste.

```python
#Ex. 7
def softmax(x):
  e_x = np.exp(x-np.max(x))
  return e_x/e_x.sum(axis=0)

vetor = softmax([2,1,0.1])
print(vetor)
print(sum(vetor))
```
